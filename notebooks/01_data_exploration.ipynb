# Digital DNA — 01: Data Exploration
Exploratory Data Analysis (EDA) of continuous behavioral biometric interaction telemetry: Keystroke dynamics and mouse pointer kinematics.

```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd

# Ensure project root is accessible
project_root = Path("..").resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from core.config import config
print("Project root:", project_root)
```

## 1. Load Keystroke and Mouse Benchmark Datasets

```python
k_df = pd.read_csv(project_root / "data" / "processed" / "benchmark_keystroke.csv")
m_df = pd.read_csv(project_root / "data" / "processed" / "benchmark_mouse.csv")

print(f"Keystroke records: {len(k_df)} (Genuine: {(k_df['label'] == 1).sum()}, Impostor: {(k_df['label'] == 0).sum()})")
print(f"Mouse records:     {len(m_df)} (Genuine: {(m_df['label'] == 1).sum()}, Impostor: {(m_df['label'] == 0).sum()})")
```

## 2. Keystroke Dynamics Summary Statistics

```python
print("=== Genuine Keystroke Metrics ===")
display(k_df[k_df["label"] == 1][["mean_dwell_time", "mean_flight_time", "words_per_minute", "rhythm_consistency", "error_rate"]].describe().T)

print("=== Impostor Keystroke Metrics ===")
display(k_df[k_df["label"] == 0][["mean_dwell_time", "mean_flight_time", "words_per_minute", "rhythm_consistency", "error_rate"]].describe().T)
```

## 3. Mouse Kinematics Summary Statistics

```python
print("=== Genuine Mouse Kinematics ===")
display(m_df[m_df["label"] == 1][["mean_speed", "mean_acceleration", "path_curvature", "direction_change_rate", "movement_entropy"]].describe().T)

print("=== Impostor Mouse Kinematics ===")
display(m_df[m_df["label"] == 0][["mean_speed", "mean_acceleration", "path_curvature", "direction_change_rate", "movement_entropy"]].describe().T)
```
